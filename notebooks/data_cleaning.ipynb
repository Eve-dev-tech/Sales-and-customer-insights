Most of our data types are string
This is good and bad.
Launch_Date and _Sales_Date should be datetime
String means data is beign interpreted as text rather than as numbers, dates,booleans or categories.thiis is often normal and common in real worl business cases, esp customer & ops data.
Sometimes a number may be interpreted as a string. In this case:(Launch_Date and Peak_Sales_Date)  are dates  that have been captured as string.

Common causes:

Cause	                      Example
CSV formatting issues	      commas, spaces, symbols
Mixed values in a column	  numbers + words
Missing values	              "N/A" or blank
Currency symbols	          "KES 1,200"
Percent signs	              "45%"
Dates not standardized	      "12/05/2026"
Human-entered data	          inconsistent formatting

Advantanges:
1. Rich Behavioral/Qualitative Information(This can reveal behavioral patterns.)
this is VERY valuable in:
customer analytics
behavioral analysis
segmentation
sentiment analysis

2.Useful for Segmentation
String columns are excellent for:
clustering customer types
identifying patterns
cohort analysis
customer journey mapping

3.Useful for NLP Later
Text data enables future expansion into:
sentiment analysis
topic modeling
customer feedback analytics
review analysis

Disadvantages
1.Models Cannot Directly Use Strings
Machine learning models require numerical input. model.fit(X, y)
We will encode them first.

2.Hidden Data Quality Problems
String columns often hide:
Hidden Issue	                  Example
inconsistent capitalization	      Male vs male
spelling errors	                  Nairobi vs Nairob
trailing spaces	                  "Yes "
mixed formatting	              "KES1200" vs "1200"
duplicate categories	          "Mobile-App" vs "Mobile App"

3.Memory Inefficiency
Strings consume much more memory than numeric types.
Large datasets become:
slower
heavier
less efficient

4.Slower Computation
Operations on strings are slower than numerical operations.
This matters for:
large datasets
dashboards
ML pipelines
forecasting systems

5.Statistical Analysis Becomes Harder
I cannot directly compute:
averages
correlations
distributions
regressions
e.g df.mean()

6.Risk of False Numerical Interpretation
e.g 100 looks numeric but behaves as text. it may cause "100"+"50"=10050 instead of 150 for dates then:

sorting may fail
time-series analysis breaks
forecasting becomes unreliable 

I will convert when required
